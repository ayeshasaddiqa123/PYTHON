```python
class Solution:
    def __init__(self) :
        self.l = []
    def enter_array(self,*args) :
        for num in args :
         self.l.append(num)
        print("You entered array : " ,self.l)

    def twoSum(self,  target: int) :
        length = len(self.l)
        
        for i in range (0,length) :
            for j in range(0,length) :
                if self.l[i] + self.l[j]  == target :
                    return[i,j]
                   
                    
        
        return []   

            

        




s = Solution()
array = []
for i in range(0,4) :
   arr = int(input("Enter elements of  array :"))
   array.append(arr)
target = int(input("Enter target :"))   
l = s.enter_array(*array)
print("List of indices is : " , s.twoSum(target))
   
```

```python
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        n = len(haystack)
        m = len(needle)
        for i in range(n-m+1) :
            if  haystack[i:i+m] == needle :
                return i
        return -1
if "__name__" == "__main__" :
  haystack = "sadbutsad"
  needle = "sad"
  s = Solution()
  print(s.strStr(haystack , needle))
```

```python
def detect(nums ,target) :
    
    
    for i in range(len(nums)) :
              if nums[i] >= target :
                return i 

    return len(nums) 
            
     
detect([1,2,3,5] ,4)     
```
