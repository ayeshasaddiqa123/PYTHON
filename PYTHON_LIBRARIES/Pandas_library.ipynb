```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')
```

```python
df=pd.read_csv('students.csv')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
categorial_cols = ["student_id" , "name" , "section" ,"gender" , "address" , "city" , "parent_name" , "pass_status" ,"division" , "exam_date"]

for col in categorial_cols:
    df[col]=df[col].astype("category")
df.dtypes
```


Univariate Analysis

```python
df.hist(figsize=(18,14) , bins=20)
plt.show()
```

```python
categorial_cols = ["student_id" , "name" , "section" ,"gender" , "address" , "city" , "parent_name" , "pass_status" ,"division" , "exam_date"]

for col in categorial_cols:
    plt.figure(figsize=(12,14))
    sns.countplot(data=df , x=col)
    plt.title(f"Countplot - {col}")
    plt.xticks(rotation=45)
    plt.show()
```

```python
plt.figure(figsize=(12,14))
sns.scatterplot(data=df , x="science_marks",y="attendance_percentage",hue="pass_status")
plt.show()
```

```python
plt.figure (figsize = (12,6))
sns.boxplot( data=df , x= 'total_marks' , y='pass_status',hue="gender")
plt.xticks( rotation = 45)
plt.show()
```

```python
plt.figure(figsize=(14,10))
sns.heatmap(df.select_dtypes(include=np.number).corr(), annot=True, cmap="coolwarm")
plt.show()
```
