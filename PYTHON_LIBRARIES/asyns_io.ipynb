```python
import time

def task1():
    print("Task 1 started")
    time.sleep(4)
    print("Task 1 finished")

def task2():
    print("Task 2 started")
    time.sleep(4)
    print("Task 2 finished")

task1()
task2()
```

```python
import asyncio



async def task1 ():
    print("Task1 satart")
    await asyncio.sleep(4)
    print("Task 1 completed")
    
    
async def task2 ():
    print("Task2 satart")
    await asyncio.sleep(4)
    print("Task 2 completed")
    
    
async def main () :
    await asyncio.gather(
        task1(),
        task2()
    )   
    
       
await main()     
    
```
