```python
import numpy as np

data_marks = np.array([45,56,78,60,94])
print("Marks of students are: ", data_marks)
print("Type of data_marks: ", type(data_marks))
```

```python
mean_score = np.mean(data_marks)
print("Mean score of data_marks (Average) : ", mean_score)  

median_score = np.median(data_marks)
print("Median score of data_marks (Middle value ) : ", median_score)


std_dev = np.std(data_marks)
print(f"Standard Deviation of data_marks : {std_dev : .2f}")

variance = np.var(data_marks)
print(f"Variance of data is : {variance :.2f}")

print("Minimum value in data_marks : ", np.min(data_marks))
print("Maximum value iin data_marks :" , np.max(data_marks))


```

```python
import pandas as pd

data_dict ={
    'Student' :['Babar' , 'Amna' ,'Alishba' , 'Wajeeha' , 'Sana'],
    'Math' : [45,56,78,60,94] ,
    'Science' : [55,66,88,70,80],
    'English' : [65,76,98,80,90],
    'study_hours' : [1,3,6,4,5]
    
}

df =pd.DataFrame(data_dict)
print(df)
```

```python
print("First three rows \n" , df.head(3),"\n\n")

print("Last two rows \n" , df.tail(2),"\n\n")

print("Data Information \n" , df.info(),"\n\n")

print("Shape of columns and rows \n" , df.shape)
```

```python
df['Average'] = df[['Math', 'Science', 'English']].mean(axis=1)

print("DataFrame with Average column \n" , df)

df_sorted = df.sort_values(by='Average', ascending=False)
print("DataFrame sorted by Average in descending order \n" , df_sorted)
```

```python
import matplotlib.pyplot as plt
plt.hist(df['Average'], bins=5, edgecolor='black')
plt.title('Distribution of Average Scores')
plt.xlabel('Average Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
#relation between study hours and average score
plt.scatter(df['study_hours'], df['Average'])
plt.title('Study Hours vs Average Score')
plt.xlabel('Study Hours')
plt.ylabel('Average Score')
plt.grid()  
plt.show()
```

```python
import seaborn as sns 
sns.set_style('whitegrid')
np.random.seed(42)

sample_data = pd.DataFrame({
    'Math' : np.random.normal(75,12,100),
    'Science' : np.random.normal(70,15,100),
    'English' : np.random.normal(80,10,100),
    'Gender' : np.random.choice(['Male', 'Female'], size=100),
    'Grades' : np.random.choice(['A', 'B', 'C', 'D'], size=100)
    
})


```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data = sample_data , x='Grades' , y='Math' ,palette = 'Set2' )

plt.title('Math score distribution by class' ,fontsize = 14 ,fontweight = 'bold')
plt.xlabel('Class')
plt.xlabel('Math score')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.violinplot(data = sample_data , x='Gender' , y='Science' ,palette = 'Set2' )
plt.title('Science score distribution by Gender' ,fontsize = 14 ,fontweight = 'bold')
plt.xlabel('Gender')
plt.ylabel('Science score')
plt.show()
```

```python
plt.figure(figsize=(10,6))
correlation_matrix = sample_data[['Math', 'Science', 'English']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Matrix of Scores')
plt.show()
```

```python
pair_plot = sns.pairplot(sample_data[['Math', 'Science', 'English' ,'Gender']] ,hue = 'Gender' ,diag_kind='hist' ,palette='husl' )
pair_plot.fig.suptitle('Pair Plot of Scores by Gender' ,fontsize = 14 ,fontweight = 'bold')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(data=sample_data, x='Grades',hue ='Gender' , palette='Set2')
plt.title('Count of Students by Grades', fontsize=14, fontweight='bold')
plt.xlabel('Grades')
plt.ylabel('Count')
plt.show()
```
