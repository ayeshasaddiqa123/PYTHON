**Problem Statement:** The mean age of a population is known to be 23 years. A researcher takes a random sample of 42 individuals and finds their average age to be 23.8 years. The population standard deviation is known to be 2.4 years. Perform a two-tailed Z-test to determine if the mean age of the population has changed, using significance levels of α = 0.05 and α = 0.02.

```python
# two tailed test
import numpy as np
from scipy.stats import norm

mu = 23
n = 42
x_bar = 23.8
std = 2.4

# calculate z statistic
z = (x_bar - mu) / (std / np.sqrt(n))
print("z statistic :" , np.round(z, 4))
print("_"*50)

# critical values for alpha = 0.05
alpha_5 = 0.05
z_critical_5 = norm.ppf(1 - alpha_5 / 2)
print("critical value for alpha=0.05 : ±" , np.round(z_critical_5, 4))
print("_"*50)

if abs(z) >= z_critical_5:
    print("Reject the null hypothesis at alpha=0.05")
else:
    print("Fail to reject the null hypothesis at alpha=0.05")
print("_"*50)    

# critical values for alpha = 0.02
alpha_2 = 0.02
z_critical_2 = norm.ppf(1 - alpha_2 / 2)
print("critical value for alpha=0.02 : ±" , np.round(z_critical_2, 4))

if abs(z) >= z_critical_2:
    print("Reject the null hypothesis at alpha=0.02")
else:
    print("Fail to reject the null hypothesis at alpha=0.02")

```

One Tail Z-test { Left-tail }
Problem Statement: A juice company claims that each bottle contains at least 500 ml of juice.

You suspect the machine is overfilling the bottles.

You take a random sample of 36 bottles and find the average juice = 506 ml.

The population standard deviation is known to be σ = 12 ml.

At α = 0.05, test whether the bottles are being overfilled.

```python
mu = 500
n = 36
x_bar = 506
sigma = 12

# calculate z statistic
z = (x_bar - mu) / (sigma / np.sqrt(n))
print("z statistic :" , np.round(z, 4))
print("_"*50)

# critical values for alpha = 0.05
alpha_5 = 0.05
z_critical_5 = norm.ppf(1 - alpha_5 / 2)
print("critical value for alpha=0.05 : ±" , np.round(z_critical_5, 4))
print("_"*50)
if abs(z) >= z_critical_5:
    print("Reject the null hypothesis at alpha=0.05")
else:
    print("Fail to reject the null hypothesis at alpha=0.05")
    
```
