```python
#  padas is used for data manipulation and analysis
import pandas as pd
import numpy as np # NumPy is used for numerical operations and handling arrays
import seaborn as sns # Seaborn is used for data visualization and statistical graphics
import matplotlib.pyplot as plt # Matplotlib is used for creating static, animated, and interactive visualizations in Python

sns.set_theme(style="whitegrid")

print("Libraries imported successfully.")
```

```python
# Setting a random seed for reproducibility
np.random.seed(42)
days = np.arange(1,101) # Simulating 100 days of sensor data
true_temperature = 50 + (days * 0.1) 
noise = np.random.normal(loc=0 ,scale =3 ,size=100) # Simulating noise with a normal distribution (mean=0, std=3)
observed_temperature = true_temperature + noise

# Creating a DataFrame to hold our simulated sensor data
df = pd.DataFrame({
    'Day' : days,
    'Temperature' : observed_temperature
})

print("5 rows of our sensor data : ")
display(df.head())
```

***Descriptive Statistic**

```python
# Statistical summary of the temperature data
summary = df['Temperature'].describe()
print("----------------Statistical Summary------------------")
print(summary)
```

```python
# Visualizing the distribution of temperature readings
plt.figure(figsize=(10,6))
sns.histplot(df['Temperature'] , kde = True , color = 'teal' ,bins =15) #kde = True adds a smooth line to show the distribution
plt.title("Distribution of temperature readings")
plt.xlabel("Temperature (°C)")
plt.ylabel("Frequency")
plt.show()

print("The highest bar is the most common temperature")
print("The smooth line shows the probability distribution")
```

```python
plt.figure(figsize  = (10,6))
sns.scatterplot(x ='Day' ,y='Temperature' , data = df , color='blue' , alpha = 0.5 ,label = 'Observed Temperature(Noise)')
# regression line to show the underlying trend
sns.regplot(x ='Day' ,y='Temperature' , data = df , scatter = False , color='red' , label = 'Trend Line(True)')
plt.title("Senspor data : Detecting  signal inside the Noise ")
plt.legend() # Add legend to differentiate between observed data and trend line
plt.show()
print("The blue dots represent the noisy observations(Messy)")
print("Red line is what the statistics sees")
print("This means the machine is getting hotter over timee")
```
