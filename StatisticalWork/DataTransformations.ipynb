```python
import pandas as pd
import numpy as np

data ={
    'Weight_in_kg' : [66.5,78.2, 90.1, 120.5, 150.0],
    'Discrete_data' :[1,2,3,4,5],
    'blood_type' : ['A', 'B', 'AB', 'O', 'A'],
    'satisfaction' :['Satisfied', 'Neutral', 'Dissatisfied', 'Satisfied', 'Neutral']
}

df = pd.DataFrame(data)

print("Sample data :")
print(df)
print("\n"+"-"*50 + "\n")


print("Data types of each column :")
print(df.dtypes)
print("\n"+"-"*50 + "\n")
print("Summary statistics for numerical columns :")
print(df.describe())
```

```python
# labelencoding is best to use for ordinal data
import sklearn 
from sklearn.preprocessing import LabelEncoder 
colors = ['Red', 'Green', 'Blue', 'Green', 'Red']
label_encoder = LabelEncoder()
encoded_colors = label_encoder.fit_transform(colors)
print("Original colors:", colors)
print("Encoded colors:", encoded_colors)
```

```python
for i  , category in enumerate(label_encoder.classes_):
    print(f"{category} -> {i}")
    
    
decoded_colors = label_encoder.inverse_transform(encoded_colors)
print("Decoded colors:", decoded_colors)    
```

```python
from sklearn.preprocessing import OneHotEncoder
vehicles = np.array(['Car', 'Bike', 'Bus', 'Car', 'Bike']).reshape(-1, 1)

print("Original vehicles:"  , vehicles.shape) #(5, 1) which is 5 rows and 1 column
print("Original Data :\n", vehicles.flatten())

one_hot_encoder = OneHotEncoder(sparse_output=False)
one_hot_encoded = one_hot_encoder.fit_transform(vehicles)
print("One-hot encoded vehicles:" , one_hot_encoded.shape) #(5, 3) which is 5 rows and 3 columns
print("One-hot Encoded Data :\n", one_hot_encoded)
categories = one_hot_encoder.categories_[0]
print("Categories:", categories)
for i, category in enumerate(categories):
    print(f"{category} -> {i}")
    
```
