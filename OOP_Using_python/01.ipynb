```python
def Sum (
         a,b,
         c,d ):
    """Sum two numbers"""
    return a+b+c+d

s= Sum(
        1,2,
        3,4
    )
print(s)
print("Documentation",\
    Sum.__doc__)
```

```python
def division(
        a,b):
    """ It returns quotient
    
    and undefined in case of divisor is 0
    """
    if (b==0 
             or a==0):
        return "Undefined"
    else :
        return a/b
    
d = division(
    10,0
) 

print(d) 
print(division.__doc__)  
```
