```python
from abc import ABC , abstractmethod
# abstract class
class Car(ABC) :
    @abstractmethod
    def start_car(self) :
        pass
    @abstractmethod
    def stop_car(self):
        pass
# subclass 1        
class Kia(Car) :
    def start_car(self) :
        print("Start  kia")
    def stop_car(self):
        print("Stop it")
# subclass 2        
class Honda(Car) :
    def start_car(self) :
        print("Start  honda")
        
    def stop_car(self): 
        print("Stop  it")    
        
kia = Kia() 
kia.start_car() 
kia.stop_car()
honda = Honda()
honda.start_car()
honda.stop_car()                 
```
