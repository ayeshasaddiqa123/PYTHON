association

```python
class Student:
    def __init__(self, name):
        self.name = name

class Teacher:
    def __init__(self, name):
        self.name = name

    def teach(self, student):
        # The teacher interacts with the student through a method
        print(f"Teacher {self.name} is teaching student {student.name}.")

# Both objects are created independently
student1 = Student("Ali")
teacher1 = Teacher("Mr. Ahmed")

# They interact via association
teacher1.teach(student1)
del student1

print(teacher1.name)
# print(student1.name)
```

aggregation

```python
class Professor:
    def __init__(self, name):
        self.name = name

class Department:
    def __init__(self, dept_name):
        self.dept_name = dept_name
        self.professors = []  # Container for Professor objects

    def add_professor(self, professor):
        self.professors.append(professor)

# 1. Create the independent objects first
prof1 = Professor("Dr. Sana")
prof2 = Professor("Dr. Bilal")

# 2. Add them to the aggregate container
cs_dept = Department("Computer Science")
cs_dept.add_professor(prof1)
cs_dept.add_professor(prof2)

# 3. If we delete the department, the professors still exist in memory!
del cs_dept
print(prof1.name)  # Output: Dr. Sana (Still exists)
```

composiition

```python
class Room:
    def __init__(self, room_type):
        self.room_type = room_type

class House:
    def __init__(self):
        # The child objects are created INSIDE the parent constructor
        self.bedroom = Room("Bedroom")
        self.kitchen = Room("Kitchen")

    def show_house(self):
        print(f"This house has a {self.bedroom.room_type} and a {self.kitchen.room_type}.")

# 1. Create the parent object
my_house = House()
my_house.show_house()

# 2. Delete the parent object
del my_house

# If you try to access my_house.bedroom now, Python throws a NameError.
# The Room objects died when the House object died.
```
