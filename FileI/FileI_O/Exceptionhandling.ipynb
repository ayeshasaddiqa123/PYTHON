```python
from pathlib import Path
import json

class UserManager:
    def __init__(self, filename="user_info.json"):
        """Initialize the manager with a file path and load existing data."""
        self.path = Path(filename)
        self.users = self._load_all_users()

    def _load_all_users(self):
        """Helper method to load the list of users from the JSON file."""
        if self.path.exists():
            try:
                contents = self.path.read_text()
                return json.loads(contents)
            except json.JSONDecodeError:
                return []
        return []

    def _save_all_users(self):
        """Helper method to write the current users list back to the JSON file."""
        self.path.write_text(json.dumps(self.users, indent=4))

    def find_user(self, name):
        """Search for a specific user profile by name (case-insensitive)."""
        for user in self.users:
            if user["Name"].lower() == name.lower():
                return user
        return None

    def register_user(self, name):
        """Prompt for remaining information, create a new user profile, and save."""
        country = input(f"What is your Country, {name}? ").strip()
        Password = input("Enter password : ")
        
        new_user = {
            "Name": name,
            "Country": country ,
            "Password" : Password
        }
        
        self.users.append(new_user)
        self._save_all_users()
        return new_user

    def delete_user(self):
        """Find and remove a specific user profile from the list and JSON file."""
        print("\n--- Delete User Profile ---")
        name_to_delete = input("Enter the Name of the user to delete: ").strip()
        
        # Look for the user object first
        matched_user = self.find_user(name_to_delete)
        
        if matched_user:
            # Remove the exact object from our in-memory list
            self.users.remove(matched_user)
            # Rewrite the updated list to the JSON file
            self._save_all_users()
            print(f" Success: User '{matched_user['Name']}' has been permanently deleted.")
        else:
            print(f" Error: No user named '{name_to_delete}' was found in the database.")

    def view_all_users(self):
        """Display a formatted list of all stored user profiles."""
        print("\n--- Registered Users Database ---")
        if not self.users:
            print("The database is currently empty.")
            return
        
        for idx, user in enumerate(self.users, start=1):
            print(f"{idx}. {user['Name']} ({user['Country']})")

    def update_user_country(self):
        """Find an existing profile and update its country value."""
        print("\n--- Update User Profile ---")
        name = input("Enter the Name of the user to update: ").strip()
        user = self.find_user(name)
        
        if user:
            new_country = input(f"Enter new Country for {user['Name']}: ").strip()
            user['Country'] = new_country
            self._save_all_users()
            print(f" Success: Updated {user['Name']}'s country to {new_country}.")
        else:
            print(f" Error: User '{name}' not found.")

    def authenticate_and_greet(self):
        """Identify, greet, or register a user."""
        print("\n--- Login / Registration ---")
        current_name = input("What is your Name? ").strip()
        
        matched_user = self.find_user(current_name)
        
        if matched_user:
            print(f"Welcome back, {matched_user['Name']} from {matched_user['Country']}!")
        else:
            print(f"Hello {current_name}, you are a new user!")
            new_user = self.register_user(current_name)
            print(f"We will remember you when you come back, {new_user['Name']}!")

    def show_menu(self):
        """Display an interactive main menu interface for the application."""
        while True:
            print("\n==============================")
            print("         MAIN MENU            ")
            print("==============================")
            print("1. Login / Greet User")
            print("2. Delete a User Profile")
            print("3. View All Registered Users")
            print("4. Update User Country")
            print("5. Exit Application")
            
            choice = input("\nSelect an option (1-5): ").strip()
            
            if choice == "1":
                self.authenticate_and_greet()
            elif choice == "2":
                self.delete_user()
            elif choice == "3":
                self.view_all_users()
            elif choice == "4":
                self.update_user_country()
            elif choice == "5":
                print("\nGoodbye! Thank you for using the system.")
                break
            else:
                print(" Invalid choice. Please type a number from 1 to 5.")

# --- Execution Entry Point ---
if __name__ == "__main__":
    manager = UserManager()
    # Launch the interactive menu loop
    manager.show_menu()
```
