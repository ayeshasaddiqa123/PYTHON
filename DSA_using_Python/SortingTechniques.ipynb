```python
# Bubble sort implementation in Python
from array import *
class BubbleSort:
    def __init__(self):
        self.array = array('i', [])
        max_size = 5

    def input_array(self) :
        for i in range(5):
            num = int(input("Enter a number: "))
            self.array.append(num)
        print("Original array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()
    def sort_array(self) :
        n = len(self.array)
        for i in range(n):
            for j in range(0, n-i-1):
                if self.array[j] > self.array[j+1]:
                    self.array[j], self.array[j+1] = self.array[j+1], self.array[j]
                    
    def display_array(self) :
        print("Sorted array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()

b = BubbleSort()
b.input_array()
b.sort_array()
b.display_array()        


        
        
```

```python
# insertion sort implementation in Python

from array import *
class InsertionSort:
    def __init__(self):
        self.array = array('i', [])
        max_size = 5

    def input_array(self) :
        for i in range(5):
            num = int(input("Enter a number: "))
            self.array.append(num)
        print("Original array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()

    def sort_array(self) :# using two for loops
        n = len(self.array)
        for i in range(1, n):
            key = self.array[i]
            j = i-1
            while j >= 0 and key < self.array[j]:
                self.array[j + 1] = self.array[j]
                j -= 1
            self.array[j + 1] = key

            
    def display_array(self) :
        print("Sorted array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()

b = InsertionSort()
b.input_array()
b.sort_array()
b.display_array()   

```

```python
#Selection sort implementation in Python

from array import *
class SelectionSort:
    def __init__(self):
        self.array = array('i', [])
        max_size = 5

    def input_array(self) :
        for i in range(5):
            num = int(input("Enter a number: "))
            self.array.append(num)
        print("Original array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()

    def sort_array(self):
        n = len(self.array)
        for i in range(n):
            min_idx = i
            for j in range(i+1, n):
                if self.array[j] < self.array[min_idx]:
                    min_idx = j
            self.array[i], self.array[min_idx] = self.array[min_idx], self.array[i]


    def display_array(self) :
        print("Sorted array is: ", end="")
        for i in range(len(self.array)):
            print(self.array[i], end=" ")
        print()

b = SelectionSort()
b.input_array()
b.sort_array()
b.display_array()  
```
