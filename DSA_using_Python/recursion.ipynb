```python
# finding factorial with loop
class Fact :
    def fact(self,n) :
        f= 1
        for i in range(1,n+1) : 
            f = f *i

        print("Factorial of ",n,"is :",f)    

        
        
    
        
        
f = Fact()
n = int(input("Enter number to find factorial :"))        
f.fact(n)

```

```python
# finding factorial with recursiuon

class Factorial :

    def Fact(self,n):

        if n==1 or n==0 :

            return 1
        return n *self.Fact(n-1)
    
f=Factorial()
n = int(input("Enter number to find factorial :"))        
print("factorial is : ", f.Fact(n))
```
